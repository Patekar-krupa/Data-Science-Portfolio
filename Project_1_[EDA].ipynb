```python
print("---"*70)
print("FINDING MISSING VALUES")
print("---"*70)
print(df.isnull().sum())  #finding missing values from the dataset using sum operation
```

Handling Missing & Duplicate Records

```python
print("---"*70)
print("Handling Missing Values")
print("---"*70)
df["Age"]=df["Age"].fillna(df["Age"].median()) #beefore replacement we had 177 missing values in Age column , so I replaced it with using statistical measure "Median".
print("---"*70)
print("Missing Values After Repplacment")
print("---"*70)
print(df.isnull().sum())
print("---"*70)
print("Finnding Duplicate Values")
print("---"*70)
duplicate_count=df.duplicated().sum()
print(f"Number of duplicated values: {duplicate_count}")
```

Grouping Age and Survival Columns

```python
df["Age"].value_counts()
df.groupby("Age")["Survived"].sum() #Age annd Survival rate
survival_rate = df.groupby("Age")["Survived"].mean() * 100
print(survival_rate)
```

Findings 1
Did women have a higher survival rate than men ?
Total men =577 & women= 314
Women survived 74% than men .

```python
#Q1 Did women have a higher survival rate than men ?
df["Sex"].value_counts()  #total men =577 & women= 314
df.groupby("Sex")["Survived"].sum()  #total survival rate men=109 & women=233
survival_rate = df.groupby("Sex")["Survived"].mean() * 100
print(survival_rate)  #about 74% women survived than men
                      #Yes. Women had higher rate of survival than men.
```

Findings 2 Which age group had the highest number of passengers ?
Young Age Group had a higher rate of passengers.

```python
#Q2 Which age group had the highest number of passengers?
df["Age"].describe()
bins = [0, 12, 19, 35, 59, 80]
labels = [
    "Children",
    "Teenagers",
    "Young Adults",
    "Adults",
    "Seniors"
]
df["Age_Group"] = pd.cut(
    df["Age"],
    bins=bins,
    labels=labels,
    include_lowest=True
)
plt.figure(figsize=(6,4))

df["Age_Group"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Age Group Distribution")
plt.ylabel("")
plt.show()
```

Findings 3 Did ticket fare and passenger class have a relationship?
The negative correlation means that as the passenger class number increases (from 1 to 3), ticket fare generally decreases.

```python
#Q3.	Did ticket fare and passenger class have a relationship?
df[["Pclass", "Fare"]].head(6)
correlation = df["Pclass"].corr(df["Fare"])
print("Correlation:", correlation) #The negative correlation means that as the passenger class number increases (from 1 to 3), ticket fare generally decreases.
```
