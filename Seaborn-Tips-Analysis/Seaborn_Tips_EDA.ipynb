```python
#Finding and replacing missing vlaues
print("=="*70)
print("Finding Missing Values")
print("=="*70)
print(df.isnull().sum())      #since there are no missing values , no replacement required.

du=df.duplicated().sum()
print(du)

print("=="*70)
print(df.describe())
print("=="*70)
```

**Q1. Which day of the week generates the highest average restaurant bill?**

```python
sns.barplot(data=df,x="day",y="total_bill")
plt.title("Day which generates the highest Average bill")
plt.xlabel("Days")
plt.ylabel("Average Bills")
plt.show()
```

Sunday generates the highest average restaurant bill compared with Thursday, Friday and Saturday in the Tips dataset.

**Q2 Do smokers and non-smokers give different average tips?**

```python
sns.barplot(data=df,x="day",y="tip",hue="smoker")
plt.xlabel("Days")
plt.ylabel("Tips")
plt.title("Do smokers and non-smokers give different average tips")
plt.show()
```

On Sunday's Smoker often offers tips .

**Q3. Which meal time (Lunch or Dinner) has more customers**

```python
meal_counts = df["time"].value_counts(normalize=True) * 100
meal_counts = meal_counts.to_frame()
sns.heatmap(meal_counts, annot=True, fmt=".1f", cmap="YlGnBu")
plt.title("Customer Count by Meal Time")
plt.xlabel("Meal Time")
plt.ylabel("")
plt.show()
```

Dinner accounts for approximately 72.1% of the records while
Lunch accounts for approximately 27.9% of the records.

**Q4. Is there a relationship between total bill and tip amount?**

```python
correlation = df["total_bill"].corr(df["tip"])
sns.scatterplot(data=df, x="total_bill", y="tip")
plt.title("Relationship between Total Bill and Tip Amount")
plt.xlabel("Total Bill")
plt.ylabel("Tip Amount")
plt.text(35, 9.5, f"Correlation: {correlation:.2f}")
plt.show()
```

There is a positive relationship between the total bill and tip amount.
Most data points are concentrated between total bills of 10–25 and tips of 1–4.
As the total bill increases, the tip amount generally increases
